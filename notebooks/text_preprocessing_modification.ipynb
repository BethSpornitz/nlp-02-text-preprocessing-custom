## Purpose
Explore how adding frequently occurring words to a stop-word list affects text preprocessing and word frequencies.

## Modification
Added "text" and "language" to the original stop-word list.

## Baseline
- Original cleaned tokens: 46
- Modified cleaned tokens: 38

```python
from collections import Counter
from pathlib import Path
import re

# Load the original example data
text = Path("../data/text_data.txt").read_text(encoding="utf-8").lower()
tokens = re.sub(r"[^a-z0-9\s]", " ", text).split()

original_stop_words = {
    "a",
    "an",
    "and",
    "are",
    "as",
    "at",
    "be",
    "but",
    "by",
    "for",
    "from",
    "has",
    "have",
    "in",
    "is",
    "it",
    "its",
    "of",
    "on",
    "or",
    "that",
    "the",
    "to",
    "was",
    "were",
    "will",
    "with",
}

modified_stop_words = original_stop_words | {"text", "language"}

original = [t for t in tokens if len(t) > 2 and t not in original_stop_words]
modified = [t for t in tokens if len(t) > 2 and t not in modified_stop_words]

print("Original cleaned tokens:", len(original))
print("Modified cleaned tokens:", len(modified))
print("Original top 5:", Counter(original).most_common(5))
print("Modified top 5:", Counter(modified).most_common(5))
```

```python
import matplotlib.pyplot as plt

labels = ["Original", "Modified"]
counts = [len(original), len(modified)]

plt.figure(figsize=(7, 4))
plt.bar(labels, counts)
plt.title("Effect of Expanding the Stop-Word List")
plt.ylabel("Number of Cleaned Tokens")
plt.ylim(0, 55)

for i, count in enumerate(counts):
    plt.text(i, count + 1, str(count), ha="center")

plt.tight_layout()
plt.show()
```

## Results and Observations

The original code produced 46 tokens. After adding "text" and "language" to the stop-word list, the number decreased
to 38, which removed 8 additional tokens.

## Insights

This modification showed that customizing stop words can affect word frequency results.

Removing common words may help highlight other terms, but sometimes you need to consider the analysis purpose it is before removing
words that could contain meaningful information.
